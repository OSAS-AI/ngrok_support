```python
import os
import subprocess
import time
from pyngrok import ngrok, conf

# 1. Authenticate pyngrok (replace 'XXX' with your actual token)
conf.get_default().auth_token = "XXX"

# 2. Configure tunnel host (optional)
tunnel_host = ""  # Set to specific host if needed, e.g., "localhost"

# 3. Build connection parameters
connect_kwargs = {
    "addr": 11434,
    "proto": "http",
    "host_header": "localhost:11434"
}

# Add bind_host if tunnel_host is specified
if tunnel_host and tunnel_host not in ["localhost", ""]:
    if not tunnel_host.startswith(("http://", "https://")):
        tunnel_host = f"http://{tunnel_host}"
    connect_kwargs["bind_host"] = tunnel_host

# 4. Start the tunnel
public_url = ngrok.connect(**connect_kwargs).public_url
print(f"Ollama is now accessible at: {public_url}")
```
